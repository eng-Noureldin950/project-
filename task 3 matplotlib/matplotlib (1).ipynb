```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
iris= sns.load_dataset('iris')
```

```python
iris
```

```python
flights = sns.load_dataset('flights')
```

```python
flights 
```

```python
annual_flights = flights.groupby('year')['passengers'].sum().reset_index()


plt.figure(figsize=(8, 5))


plt.plot(annual_flights['year'], annual_flights['passengers'], color='red', marker='o', linewidth=2)


plt.title('Total Passengers Per Year (1949 - 1960)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Passengers', fontsize=12)
plt.grid(True, linestyle='-', alpha=0.6) 


plt.show()
```

the number of passenger increase continously over the years 

```python
tips= sns.load_dataset('tips')
```

```python
tips 
```

Step 1: Line & Scatter Plots

```python
plt.figure(figsize=(8, 5)) 
plt.scatter(tips['total_bill'], tips['tip'], color='blue', alpha=0.5, marker='o')


plt.title('Total Bill vs Tip (Matplotlib)') 
plt.xlabel('Total Bill ($)') 
plt.ylabel('Tip ($)') 
plt.grid(True)
plt.show()
##using matplotlib 
# ---------------------------------------------------

## using Seaborn 
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', palette='Set1')

plt.title('Total Bill vs Tip (Seaborn)')
plt.show()
```

THE more total_bill , the MORE TIPS

# step 2 &3 styling and customization 

```python
# ========================================================
# Step 2: Styling & Customization 
# ========================================================
sns.set_theme(style="darkgrid", palette="muted") 

# ========================================================
# Step 3: Distribution Plots 
# ========================================================

plt.figure(figsize=(9, 5))
sns.histplot(data=tips, x='total_bill', kde=True, color='purple', bins=20)

plt.title('Distribution of Total Bills (Tips Dataset)', fontsize=14, fontweight='bold')
plt.xlabel('Total Bill ($)')

plt.show()

# --------------------------------------------------------

## using seaborn 
plt.figure(figsize=(9, 5))
sns.ecdfplot(data=iris, x='sepal_length', hue='species', linewidth=2.5)

plt.title('ECDF Plot of Sepal Length by Species (Iris Dataset)', fontsize=14, fontweight='bold')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Proportion')
plt.show()
```

Step 4 :categorical and comparison 

```python

sns.set_theme(style="whitegrid")

## BOX plot 
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', hue='day', palette='Set2', legend=False)
plt.title('Comparing Total Bills Across Days (Box Plot )', fontsize=13, fontweight='bold')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()

# ----------------------------------------------------------------------

## Violin plot 
plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x='time', y='tip', hue='sex', split=True, palette='pastel')
plt.title('Tips Comparison: Lunch vs Dinner by Sex (Violin Plot)', fontsize=13, fontweight='bold')
plt.xlabel('Time of Day')
plt.ylabel('Tip ($)')
plt.show()

# ----------------------------------------------------------------------

## Bar chart 
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x='smoker', y='total_bill', hue='sex', errorbar=None, palette='muted')
plt.title('Average Total Bill: Smokers vs Non-Smokers', fontsize=13, fontweight='bold')
plt.xlabel('Smoker Status')
plt.ylabel('Average Total Bill ($)')
plt.show()
```

```python
iris_numeric = iris.drop(columns=['species'])
iris_corr = iris_numeric.corr()

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(data=iris_corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, ax=ax)

ax.set_title('Iris Features Correlation (Heatmap)', fontsize=14, fontweight='bold', color='darkred')
plt.show()
```

5) Heat map and subplot 

```python
iris_numeric = iris.drop(columns=['species'])
iris_corr = iris_numeric.corr()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.heatmap(data=iris_corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, ax=axes[0])
axes[0].set_title('1. Iris Features Correlation (Heatmap)', fontsize=13, fontweight='bold', color='darkred')

sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue='species', palette='Set1', s=70, ax=axes[1])
axes[1].set_title('2. Petal Length vs Petal sWidth by Species', fontsize=13, fontweight='bold', color='darkred')
axes[1].set_xlabel('Petal Length (cm)')
axes[1].set_ylabel('Petal Width (cm)')

plt.tight_layout()
plt.show()
```
